```python
import pandas as pd

file1 = "datahashtagreels.csv"
file2 = "datahashtagposts.csv"
file3 = "datasearchpopular.csv"
output_file = "combined_urls.csv"

df1 = pd.read_csv(file1, usecols=["url"], encoding="utf-8-sig")
df2 = pd.read_csv(file2, usecols=["url"], encoding="utf-8-sig")
df3 = pd.read_csv(file3, usecols=["url"], encoding="utf-8-sig")

combined = pd.concat([df1, df2, df3], ignore_index=True)

# Optional: clean blanks and remove duplicate URLs
combined["url"] = combined["url"].astype(str).str.strip()
combined = combined[combined["url"] != ""].drop_duplicates(subset=["url"], keep="first")

combined.to_csv(output_file, index=False, encoding="utf-8-sig")
print(f"Saved {len(combined)} URLs to {output_file}")
```

```python
from apify_client import ApifyClient
import pandas as pd
import os

# Initialize the ApifyClient with your API token
client = ApifyClient("apify_api_ieBDb5QrDKJAM3mii5BhTlUgBJ4c3v0OZVTh")

urls_df = pd.read_csv("combined_urls.csv", usecols=["url"], encoding="utf-8-sig")

direct_urls = (
    urls_df["url"]
    .dropna()
    .astype(str)
    .str.strip()
    .loc[lambda s: (s != "") & (s.str.lower() != "nan")]
    .drop_duplicates()
    .tolist()
)

# Prepare the Actor input
run_input = {
    "resultsType": "comments",
    "directUrls": direct_urls,
    "resultsLimit": 100
}

# Run the Actor and wait for it to finish
run = client.actor("shu8hvrXbJbY3Eb9W").call(run_input=run_input)
# Fetch and print Actor results from the run's dataset (if there are any)
for item in client.dataset(run.default_dataset_id).iterate_items():
    print(item)

comments_data = []
for item in client.dataset(run.default_dataset_id).iterate_items():
    comments_data.append(
        {
            "post_url": item.get("postUrl") or item.get("inputUrl"),
            "comment_id": item.get("id"),
            "username": item.get("ownerUsername") or item.get("username"),
            "text": item.get("text"),
            "timestamp": item.get("timestamp"),
            "likes_count": item.get("likesCount"),
            "replies_count": item.get("repliesCount"),
        }
    )

new_df = pd.DataFrame(comments_data)
csv_path = "l.csv"

# Concat new rows below existing CSV rows if the file already exists
if os.path.exists(csv_path) and os.path.getsize(csv_path) > 0:
    existing_df = pd.read_csv(csv_path, encoding="utf-8-sig")
    df = pd.concat([existing_df, new_df], ignore_index=True)
else:
    df = new_df

# Optional: drop empty texts and duplicate comments
df = df.dropna(subset=["text"])
df["text"] = df["text"].astype(str).str.strip()
df = df[df["text"] != ""].drop_duplicates(subset=["comment_id", "post_url", "username", "text"])

df.to_csv(csv_path, index=False, encoding="utf-8-sig")
print(f"Total rows in CSV: {len(df)} (new rows fetched: {len(new_df)})")

```

```python
import pandas as pd

csv1 = "l.csv"
csv2 = "sensus_ekonomi_2026_ig_comments.csv"
csv3 = "p.csv"
output_file = "final_ig_comments.csv"

# Read posts_url from csv1 and rename it to 'url'
df1 = pd.read_csv(csv1, usecols=["post_url", "username", "text", "likes_count", "timestamp"], encoding="utf-8-sig").rename(
    columns={"post_url": "url", "username": "username", "text": "text", "likes_count": "likes_count", "timestamp":"timestamp"}
)

# Read url from csv2
df2 = pd.read_csv(csv2, usecols=["post_url", "username", "text", "likes_count", "timestamp"], encoding="utf-8-sig").rename(
    columns={"post_url": "url", "username": "username", "text": "text", "likes_count": "likes_count", "timestamp":"timestamp"}
)

df3 = pd.read_csv(csv3, usecols=["url", "ownerUsername", "text", "likesCount", "timestamp"], encoding="utf-8-sig").rename(
    columns={"url": "url", "ownerUsername": "username", "text": "text", "likesCount": "likes_count", "timestamp":"timestamp"}
)

# Stack df2 below df1 (or [df2, df1] if you want csv2 first)
combined = pd.concat([df1, df2, df3], ignore_index=True)

# Clean blanks/NaN strings and deduplicate by 'url'
url_col = "url" if "url" in combined.columns else "post_url"

for col in [url_col, "username", "text"]:
    if col in combined.columns:
        combined[col] = combined[col].fillna("").astype(str).str.strip()

combined = combined[
    (combined[url_col] != "") & (combined[url_col].str.lower() != "nan")
]

subset_cols = [c for c in [url_col, "username", "text"] if c in combined.columns]
combined = combined.drop_duplicates(subset=subset_cols, keep="first")

combined.to_csv(output_file, index=False, encoding="utf-8-sig")
print(f"Saved {len(combined)} unique URLs to {output_file}")
```
